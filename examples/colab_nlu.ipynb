```python
from transformers import pipeline

pipe = pipeline("text-classification", model="Wouze/laya-ara", trust_remote_code=True)

out = pipe.model.predict(
    {"message": "الحوالة ما وصلت، أبي استرجاع وإلا بنقلع"},
    {
        "queue": {
            "type": "choice",
            "instructions": "Support queue",
            "criteria": {
                "billing": "payments, refunds",
                "technical": "bugs, outages",
                "other": "none of the above",
            },
        },
        "refund": {"type": "noul", "instructions": "Asks for a refund?"},
    },
)
print(out["answers"])
```
